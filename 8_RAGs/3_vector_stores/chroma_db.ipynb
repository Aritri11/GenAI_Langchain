```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document


embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

#create documents

doc1=Document(
        page_content="Artificial Intelligence is transforming healthcare by enabling early disease detection and personalized treatments.",
        metadata={"source": "article_ai_healthcare", "topic": "AI", "year": 2024}
    )
doc2=Document(
        page_content="Climate change is causing rising global temperatures, melting glaciers, and more frequent extreme weather events.",
        metadata={"source": "climate_report", "topic": "Environment", "year": 2023}
    )

doc3=Document(
        page_content="Quantum computing leverages quantum mechanics to perform computations much faster than classical computers for certain problems.",
        metadata={"source": "quantum_blog", "topic": "Technology", "year": 2024}
    )

doc4= Document(
        page_content="A balanced diet and regular exercise are essential for maintaining good physical and mental health.",
        metadata={"source": "health_magazine", "topic": "Health", "year": 2022}
    )

doc5=Document(
        page_content="Space exploration helps scientists understand the origins of the universe and search for extraterrestrial life.",
        metadata={"source": "space_journal", "topic": "Space", "year": 2025}
    )

vector_store=Chroma(
    embedding_function=embedding,
    persist_directory='chroma_db',
    collection_name='sample'

)
docs=[doc1,doc2,doc3,doc4,doc5]
```

```python
#adding documents to vector_store
print(vector_store.add_documents(docs))

#view documents
print(vector_store.get(include=['embeddings','documents','metadatas']))
```

```python
#search documents
vector_store.similarity_search(
    query='What is causing global rise in temperatures?',
    k=1
)
```

```python
#search with similarity score
vector_store.similarity_search_with_score(
    query='What is causing global rise in temperatures?',
    k=1
)
```

```python
#filtering of metadata
vector_store.similarity_search_with_score(
    query='',
    filter={'topic':'Space'}
)
```

```python
#update documents
updated_doc1=Document(
        page_content="Artificial Intelligence is transforming IT sectors by enabling automated systems.",
        metadata={"source": "article_ai_it", "topic": "AI", "year": 2024}
    )
vector_store.update_document(document_id='92aa91ba-4617-4545-9056-4342bc918782', document=updated_doc1)
```
