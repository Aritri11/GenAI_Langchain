```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document

embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

#create documents

documents=[Document(
        page_content="Artificial Intelligence is transforming healthcare by enabling early disease detection and personalized treatments."),Document(
        page_content="Climate change is causing rising global temperatures, melting glaciers, and more frequent extreme weather events."), Document(
        page_content="Quantum computing leverages quantum mechanics to perform computations much faster than classical computers for certain problems."), Document(
        page_content="A balanced diet and regular exercise are essential for maintaining good physical and mental health.")

]
```

```python
#create the FAISS vector store from documents
vector_store=FAISS.from_documents(
    documents=documents,
    embedding=embedding
)
```

```python
#enable MMR in the retriever

retriever=vector_store.as_retriever(
    search_type='mmr',
    search_kwargs={'k':3,'lambda_mult':1} #k=top results, lambda_mult=relevance-diversity balance
)

query='What is causing global rise in temperatures?'
results=retriever.invoke(query)
```

```python
#print retrieved content
for i,doc in enumerate(results):
    print(f"\n ----Result {i+1}---- ")
    print(f"Content:\n{doc.page_content}---")
```
