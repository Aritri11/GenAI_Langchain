```python
#For chat conversations
from langchain_ollama import ChatOllama

#Initialize the model
llm=ChatOllama(
    model='llama3.2:1b',
    temperature=0
)

#simple invocation
response=llm.invoke("Explain the concept of quantum mechanics in one line")
print(response.content)
```

```python
#For Non-chat execution
from langchain_ollama import OllamaLLM

llm= OllamaLLM(
    model='llama3.2:1b'
)
response=llm.invoke('The capital of India is')
print(response)
```

```python
from langchain_ollama import OllamaEmbeddings
embedding=OllamaEmbeddings(
    model='qwen3-embedding:4b'
)
#embed a single query
query_result=embedding.embed_query('What is langchain?')
print(query_result)
```

```python
#embed multiple documents
doc_results = embedding.embed_documents(['Document 1 content','Document 2 content'])
print (f"Embedding length: {len(doc_results)}")
```
